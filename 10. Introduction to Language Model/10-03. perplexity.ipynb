## 1. 퍼플렉시티(Perplexity)
- 문장의 길이로 정규화된 문장 확률(문장을 구성하는 토큰 확률의 프로덕트)의 역수
- 언어 모델의 평가 방법(Evaluation Metric)임
- 낮을 수록 성능이 더 좋음

$$
\begin{align*}
\text{PPL}(Wn) &= P(\bigcap_{k=1}^n w_k)^{-\frac{1}{n}}\\
&= \sqrt[n]{\frac{1}{P(\bigcap_{k=1}^n w_k)}}\\
&= \sqrt[n]{\frac{1}{\prod_{i=1}^n P(w_i | \bigcap_{j=1}^{i-1} w_{j})}}\\
&= \sqrt[n]{\frac{1}{\prod_{i=1}^n P(w_i | W_{i-1})}}
\end{align*}
$$

## 2. 분기 계수(Branching factor)
- PPL은 이 언어 모델이 평균적으로 몇 개의 선택지를 고민하고 있는지를 의미
- PPL이 n인 경우, 다음 단어를 예측하는 모든 시점(time step)마다 평균 n개의 단어를 고려함
