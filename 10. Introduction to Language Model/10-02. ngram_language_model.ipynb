## 1. SLM 모델의 한계
- 훈련 코퍼스에 확률을 계산하고 싶은 문장이나 단어가 없을 수 있음
- 확률을 계산하고 싶은 문장이 길어질수록 그 확률이 올라감
- 이를 희소 문제(Sparsity Problem)라 함

## 2. n-gram 모델의 접근 방법
- 희소 문제를 해소하기 위해 단어 시퀀스의 단어 n개 전체가 아닌 일부만 고려하는 방식을 시도
- 이는 아래의 가정에 기반함

$$
P(w_k|W_{k-1}) \sim P(w_k|\bigcap_{a=k-n}^{k-1} w_a)\\
$$

## 3. n-gram 모델의 한계
- 자연어의 문맥은 문장 전체에 걸쳐 분포할 수 있으나, 단어간의 거리가 n 이상일 경우 고려하지 못함
- 희소 문제에서 완전히 벗어날 수 없음
- n을 선택할 때 희소문제와 일반화 사이의 트레이드 오프가 존재함
