## 1. 합성곱 신경망의 대두
- 이미지 처리에 탁월한 성능을 보이는 신경망
- 다층 퍼셉트론은 몇 가지 픽셀만 값이 달라져도 민감하게 예측에 영향을 받는다는 단점이 있음
    - 1차원 텐서에서는 공간적인 구조(spatial structure) 정보가 유실됨
    - 공간적인 구조 정보: 픽셀간의 거리에 따른 연관성 등
- 이미지의 공간적인 구조 정보를 보존하면서 학습할 수 있는 방법이 필요: CNN의 탄생

## 2. 채널
- 이미지는 (높이, 너비, 채널)이라는 3차원 텐서
    - 높이: 세로 방향 픽셀 수
    - 너비: 이미지의 가로 방향 픽셀 수
    - 채널: 색 성분(흑백일 경우 1, 컬러일 경우 색상 채널 수)



## 3. 합성곱 연산(Convolution operation)
- `커널(필터)` 텐서로 이미지 위를 훑으며 아다마르 곱`Hadamard Product`하여 특성 맵`Feature map`을 반환



- 커널의 크기 `n*m`는 하이퍼파라미터로 결정
- 커널의 이동 거리 `Stride` 역시 하이퍼파라미터로 결정

## 4. 패딩(Padding)
- 커널이 이미지 위를 움직여야 하므로, 특성 맵은 이미지의 크기보다 작음
- 커널 맵이 크거나, 스트라이드가 크거나, 컨볼루션 연산이 많이 중첩될 경우, 특성 맵이 많이 작아질 수 있음
- 이를 막기 위해 이미지의 바깥에 값을 추가할 수 있는데, 이를 패딩(Padding)이라고 함



## 5. 가중치와 편향
- 커널의 각 원소가 가중치가 됨: `3*3` 커널의 경우 가중치는 9개
- 아다마르 곱 이후 동일한 편향을 각각의 원소에 더함

## 6. 특성 맵의 크기
$$ O = \lfloor\frac{I - K + 2P}{S} + 1\rfloor $$

## 7. 고차원 텐서의 합성곱 연산
- 이미지의 채널과 커널의 채널은 같은 수여야 함
- 채널마다 합성곱 연산을 수행하고 결과를 더함
- 각 채널의 값이 더해지므로 반환되는 특성 맵의 채널은 하나가 됨
- 특성맵의 채널 크기를 늘리고 싶다면 사용하는 커널의 수를 늘려야 함

## 8. 풀링(Pooling)
- 특성 맵을 다운샘플링하여 특성 맵의 크기를 줄이는 연산
- 일반적으로 최대값 풀링(Max Pooling) 또는 평균값 풀링(Average Pooling)을 사용
- 가중치는 없으나, 커널 크기와 스트라이드 등의 속성은 존재
