## 2. biLM(Bidirectional Language Model)
> 주의: biLM은 양방향 RNN이 아니라 서로 독립된 양방향의 언어모델 쌍을 말한다!
- RNN 언어 모델은 문장으로부터 단어 단위로 입력을 받음
- RNN 내부의 은닉 상태 $h_t$는 시점(time step)이 지날수록 점점 업데이트됨
- 결과적으로 RNN의 $h_t$ 값이 문장의 문맥 정보를 점차적으로 반영하게 됨
- 각 시점의 입력 벡터는 합성곱 신경망을 이용한 문자 임베딩(character embedding) 결과

## 3. biLM의 활용


- `Play`를 임베딩할 때, ELMo는 해당 시점(time step)의 모든 출력값을 연결(concat)해서 사용



- 이렇게 완성된 벡터를 ELMo 표현(representation)이라고 함
- ELMo 표현을 GloVe 임베딩 벡터와 연결(concatenate)해서 입력으로 사용
- biLM의 가중치는 고정시키고, 출력값을 가중합하는 가중치만 학습시킴
