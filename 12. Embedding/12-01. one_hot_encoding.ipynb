- 인코딩: 주어진 데이터를 정해진 규칙(code)에 따라 변환하는 것
  - 일반적으로, 정제되지 않은 데이터를 컴퓨터가 다루기 쉬운 형태로 변환하는 것
  - 언어모델의 관점에서, 주어진 자연어의 단어 집합을 인코딩하는 것이 중요함
- 단어 집합: 서로 다른 단어들의 집합(의미론적 구분이 아닌 형태적 구분)

## 1. 원-핫 인코딩(One-hot encoding)이란?
- 단어 집합의 단어를 아래와 같이 벡터로 표현한 것
  - 벡터의 차원: 단어 집합의 크기
  - 단어 집합에서 단어의 인덱스에 해당하는 차원의 벡터값은 1
  - 그 외의 모든 차원의 벡터값은 0

## 2. 한국어 문장 원-핫 인코딩 예제

```python
from konlpy.tag import Okt
okt = Okt()
token = okt.morphs("나는 자연어 처리를 배운다")
print(token)
```

```python
word2index = {}
for voca in token:
     if voca not in word2index.keys():
       word2index[voca] = len(word2index)
print(word2index)
```

```python
def one_hot_encoding(word, word2index):
    one_hot_vector = [0]*(len(word2index))
    index = word2index[word]
    one_hot_vector[index] = 1
    return one_hot_vector
```

```python
one_hot_encoding("자연어",word2index)
```

## 3. 원-핫 인코딩의 한계
- 단어의 개수가 늘어날수록 벡터의 차원이 늘어남
  - $n$개의 단어를 표현한 $n$개의 정수를 저장하기 위해 $n^n$ 크기의 벡터가 필요함
  - 각각의 벡터는 단 하나의 $1$과 나머지의 $0$으로 구성되어 효율성이 극히 낮음
- 단어의 유사도를 표현하지 못함
  - 원-핫 인코딩은 [각각의 대상 사이의 오차를 동일하게 구성함](../05.%20Softmax%20Regression/05-01.%20one-hot%20encoding.ipynb)
  - 이는 서로 독립된 대상을 다룰 때에는 훌륭하나, 독립적이지 않은 대상을 다룰 때에는 부적합함
  - 자연어의 단어는 서로 유사하거나 상이한 관계가 있어 독립적이지 않음
