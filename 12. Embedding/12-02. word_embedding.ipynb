- 인코딩 (Encoding)
  - 주어진 데이터를 컴퓨터가 처리 가능한 형식으로 변환하는 `규칙 기반(Rule-based)` 작업
  - 표현 특징:
    - 데이터 간의 의미적 관계를 반영하지 않기 위해 벡터를 독립적(Independent)으로 만듦
    - 일반적으로 고차원의 희소 벡터(Sparse Vector) 형태로 표현됨 (예: 원-핫 인코딩)
    - `SVD` 등의 기법으로 차원을 축소하여 밀집 벡터로 변환할 수 있음
- 임베딩 (Embedding)
  - 인코딩의 한 종류로, 특히 머신러닝 학습을 목표로 하는 `학습 기반(Learning-based)` 작업
  - 표현 특징:
    - 고차원 희소 벡터가 아닌 저차원의 밀집 벡터(Dense Vector)로 변환
    - 변환된 벡터 간의 유사성(Similarity)이 나타남
    - 이를 실제 데이터 간의 의미적 관계(Semantic Relationship)와 일치하도록 학습시킴

## 1. 희소 표현(Sparse Representation)
- 벡터, 행렬, 또는 텐서의 값이 대부분 0으로 표현되는 방식을 `희소 표현`이라고 함
- 희소 표현을 자연어 단어 집합에 적용할 경우 아래와 같은 한계점이 명확함

### 1.1 공간적 비효율성
- 단어의 개수가 늘어날수록 벡터의 차원이 늘어남
  - $n$개의 단어를 표현한 $n$개의 정수를 저장하기 위해 $n^n$ 크기의 벡터가 필요함
  - 각각의 벡터는 단 하나의 $1$과 나머지의 $0$으로 구성되어 효율성이 낮음

### 1.2 벡터 유사성 활용 불가
- 인코딩된 벡터는 서로 독립적인 관계를 가지며, 데이터간의 유사성을 표현하지 않음
- 이는 자연어 단어가 가진 의미 관계가 표현되지 않는다는 것을 뜻함

```python
import torch
```

```python
# 원-핫 벡터 생성
dog = torch.FloatTensor([1, 0, 0, 0, 0])
cat = torch.FloatTensor([0, 1, 0, 0, 0])
computer = torch.FloatTensor([0, 0, 1, 0, 0])
netbook = torch.FloatTensor([0, 0, 0, 1, 0])
book = torch.FloatTensor([0, 0, 0, 0, 1])
```

```python
print(torch.cosine_similarity(dog, cat, dim=0))
print(torch.cosine_similarity(cat, computer, dim=0))
print(torch.cosine_similarity(computer, netbook, dim=0))
print(torch.cosine_similarity(netbook, book, dim=0))
```

## 2. 밀집 표현(Dense Representation)
- 벡터, 행렬, 또는 텐서의 값이 대부분 0이 아닌 값으로 방식을 `밀집 표현`이라고 함
- 일반적으로 벡터 차원은 데이터 특성과 독립되며, 하이퍼파라미터 값으로 사용자가 설정함
- 변환된 벡터가 상호 독립적이지 않고 관계성을 갖게 됨
- 자연어 단어 집합을 밀집 표현으로 나타내는 것을 워드 임베딩(Word Embedding)이라고 함
