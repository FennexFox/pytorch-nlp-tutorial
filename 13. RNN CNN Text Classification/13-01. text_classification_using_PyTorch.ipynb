## 1. 훈련 데이터에 대한 이해
- 텍스트 분류 작업은 지도 학습(Supervised Learning)
- 훈련 데이터는 레이블이라는 이름의 미리 정답이 적혀있는 데이터로 구성


## 2. 훈련 데이터와 테스트 데이터
- 전체 데이터를 전부 훈련에 사용하는 것 보다는 테스트용으로 일부는 남겨놓는 것이 바람직
- 2,000개의 샘플에도 레이블이 있으므로 모델이 실제로 정답을 얼마나 맞추는지 정답률을 계산
- 텍스트 데이터의 열을 X, 레이블 데이터의 열을 y라고 명명
- 훈련 데이터(X_train, y_train)와 테스트 데이터(X_test, y_test)로 분리
- X_train과 y_train을 학습하고, X_test에 대해서 레이블을 예측하여 학습
- 모델이 예측한 레이블과 y_test를 비교해서 정답률을 계산

## 3. 단어에 대한 인덱스 부여
- nn.Embedding()은 단어 각각에 대해 정수가 맵핑된 입력에 대해서 임베딩 작업 가능
  - 단어를 빈도수 순대로 정렬하고 순차적으로 인덱스를 부여하는 방법 사용
  - 등장 빈도대로 인덱스가 부여되어, 빈도수 기반 전처리가 용이

## 5. RNN의 다-대-일(Many-to-One) 문제
- 텍스트 분류는 RNN의 다-대-일(Many-to-One) 문제
- 모든 시점(time step)에 대해서 입력을 받지만, 최종 시점의 RNN 셀만이 은닉 상태를 출력



인풋은 $x_i$부터 $x_t$까지 들어오나, 출력은 $y_t$만!
