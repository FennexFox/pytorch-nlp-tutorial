## 3. 가중치 규제 적용하기
모델의 가중치가 클 때에 비용을 추가하여 가중치가 작아지도록 유도함
- L1 규제 : 가중치 w들의 절대값 합계를 비용 함수에 추가 (L1 노름)
- L2 규제 : 모든 가중치 w들의 제곱합을 비용 함수에 추가 (L2 노름)

L1 규제는 L2 규제보다 특정 가중치를 0으로 만드는 경향이 더 큼 - 어떤 특성이 모델에 영향을 주는지 선별적으로 규제하고 싶을 때 사용
이러한 목적이 없다면 경험적으로 L2 규제가 더 잘 작동함
