- 기울기 소실(Gradient Vanishing): 역전파 과정에서 입력층으로 올라갈수록 기울기가 작아지는 현상
- 기울기 폭주(Gradient Exploding): 역전파 과정에서 입력층으로 올라갈수록 기울기가 발산하는 현상

## 1. 적절한 활성화함수 사용
- 은닉층에서는 ReLu 및 그 변종 등 아래의 조건을 만족하는 활성화함수를 사용
    - 정의역의 (거의) 모든 부분에서 기울기가 0으로 수렴하지 않음
    - 계산과 미분이 쉬워 순전파와 역전파에 계산 부하가 크게 걸리지 않음

## 2. 가중치 초기화
- 가중치 값을 어떻게 초기화하느냐에 따라 모델 훈련의 결과가 달라지기도 함(Local Minima 문제)

### 2.1 [세이비어 초기화(Xavier Initialization)]( http://proceedings.mlr.press/v9/glorot10a/glorot10a.pdf)
> **S자 형태의 활성화 함수와 함께 사용할 경우에 좋은 성능을 보임**<br>
> **ReLU와 함께 사용할 경우에는 성능이 좋지 않음**

$N_{in}$과 $N_{out}$이 각각 해당 층의 입력 뉴런, 출력 뉴런의 개수라고 할 때,
- 균등분포로 초기화할 경우

$$ W \sim U(-\sqrt{\frac{6}{N_{in}+N_{out}}},\sqrt{\frac{6}{N_{in}+N_{out}}}) $$

- 정규분포로 초기화할 경우

$$ W \sim N(0, \sqrt{\frac{2}{N_{in}+N_{out}}}) $$

- 여러 층의 기울기 분산 사이에 균형을 맞춰서 특정 층의 과대 또는 과소 표현을 막음.

### 2.2 [He 초기화(He Initialization)](https://www.cv-foundation.org/openaccess/content_iccv_2015/papers/He_Delving_Deep_into_ICCV_2015_paper.pdf)
> **딥러닝 프레임워크는 ReLU 계열의 레이어를 선언할 때, 내부적으로 이미 He Initialization을 적용**

$N_{in}$이 해당 층의 입력 뉴런의 개수라고 할 때,
- 균등분포로 초기화할 경우

$$ W \sim U(-\sqrt{\frac{6}{N_{in}}},\sqrt{\frac{6}{N_{in}}}) $$

- 정규분포로 초기화할 경우

$$ W \sim N(0, \sqrt{\frac{2}{N_{in}}}) $$

## 3. 배치 정규화(Batch Normalization)
- 인공 신경망의 각 층에 들어가는 입력을 평균과 분산으로 정규화하는 것

### 3.1 내부 공변량 변화(Internal Covariate Shift)
- 학습 과정에서 층 별로 입력 데이터 분포가 달라지는 현상
- 이전 층의 가중치 값이 바뀌게 되면, 전달되는 입력 데이터의 분포와 학습했던 시점의 분포가 차이가 발생

### 3.2 배치 정규화(Batch Normalization)
> 한 번에 들어오는 배치 단위로 정규화하는 것으로, 각 층에서 활성화 함수를 통과하기 전에 수행
- 미니 배치의 입력값에 대해 정규화한 뒤, 정규화된 입력값에 `스케일링`과 `시프팅`을 함
- `스케일링`과 `시프팅`의 값은 학습의 대상
- 정규화에 사용되는 이동평균과 분산은 학습시에만 계산하며, 테스트시에는 학습시의 값을 사용

### 3.3 층 정규화(Layer Normalization)
> 각 시점마다 입력이 서로 다른 통계치를 갖는 RNN에는 배치 정규화가 아닌 다른 방법을 사용함<br>
> 배치 단위가 아닌 특성 단위로 정규화를 실행함
