## 1. 퍼셉트론(Perceptron)
- 프랑크 로젠블라트(Frank Rosenblatt)가 1957년에 제안한 초기 형태의 인공 신경망
- 가중치와 곱해진 다수의 입력의 합이 임계치(편향)를 넘으면 결과를 내보내는 알고리즘
- 출력된 결과값은 활성화 함수를 거쳐 변형됨

## 2. 단층 퍼셉트론(Single-Layer Perceptron)
- 단층 퍼셉트론은 하나씩의 입력층과 출력층으로 구성됨
- 단층 퍼셉트론은 AND 게이트, NAND 게이트, OR 게이트 또한 구현할 수 있음
- 그러나 XOR 게이트는 구현하지 못함
- 직선 하나로 두 영역을 나눌 수 있는 문제에 대해서만 구현이 가능하기 때문



```python
def AND_gate(x1, x2):
    w1=0.5
    w2=0.5
    b=-0.7
    result = x1*w1 + x2*w2 + b
    if result <= 0:
        return 0
    else:
        return 1
```

```python
AND_gate(0, 0), AND_gate(0, 1), AND_gate(1, 0), AND_gate(1, 1)
```

```python
def NAND_gate(x1, x2):
    w1=-0.5
    w2=-0.5
    b=0.7
    result = x1*w1 + x2*w2 + b
    if result <= 0:
        return 0
    else:
        return 1
```

```python
NAND_gate(0, 0), NAND_gate(0, 1), NAND_gate(1, 0), NAND_gate(1, 1)
```

```python
def OR_gate(x1, x2):
    w1=0.6
    w2=0.6
    b=-0.5
    result = x1*w1 + x2*w2 + b
    if result <= 0:
        return 0
    else:
        return 1
```

```python
OR_gate(0, 0), OR_gate(0, 1), OR_gate(1, 0), OR_gate(1, 1)
```

## 3. 다층 퍼셉트론(MultiLayer Perceptron, MLP)
- 다층 퍼셉트론은 입력층과 출력층 사이에 층이 추가됨
- 입력층과 출력층 사이에 존재하는 층을 은닉층(hidden layer)이라고 함
- 이를 통해 XOR 게이트를 구현할 수 있음



- 은닉층이 2개 이상인 신경망을 심층 신경망(Deep Neural Network, DNN)이라고 함
