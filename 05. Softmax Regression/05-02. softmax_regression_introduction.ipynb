## 1. 다중 클래스 분류(Multi-Class Classification)
- 분류 작업에 주어진 레이블이 둘 초과일 경우, 다중 클래스 분류라고 부른다.
- 이진 분류 작업과 달리, 각 범주에 0부터 1 사이의 확률값을 부여하며, 그 총합은 1이 되어야 한다.
- 즉, 소프트맥스 회귀는 선택지의 개수만큼의 차원을 갖는 벡터를 반환하며, 그 벡터의 합은 1이 되어야 한다.

## 2. 소프트맥스 함수(Softmax function)
$$ \text{Softmax}(i) = p_{i} = \frac{e^{z_i}}{\sum\limits_{j=1}^{k} e^{z_j}} $$
$$ \sum\limits_{i=1}^{k} \text{Softmax}(i) = \frac{\sum\limits_{j=1}^{k} e^{z_j}}{\sum\limits_{i=1}^{k} e^{z_i}} = 1 $$
- 소프트맥스 함수는 아래와 같은 특징을 가진다.
  - 실수 전체를 정의역으로 가지므로 모든 범위의 입력값을 받을 수 있다.
  - 볼록함수인 지수함수로 구성되어 있어, 입력값의 작은 차이를 확률값의 큰 차이로 반환한다.
  - $k$차원의 벡터가 주어질 때, 해당 벡터의 모든 값은 0부터 1 사이의 값을 가지며, 벡터의 총합은 1이 된다.

## 4. 비용 함수
- 소프트맥스 회귀는 각 샘플의 정답 레이블 분포와 모델이 예측한 확률 분포 사이의 차이를 측정하기 위해 크로스 엔트로피(Cross-Entropy)를 사용한다.
- $y$를 정답의 원-핫 벡터, $\hat{y}$를 모델의 예측 확률분포라고 하면, 한 샘플에 대한 크로스 엔트로피는 다음과 같다.
  $$ \mathcal{L}(y, \hat{y}) = - \sum_{i=1}^{k} y_i \log(\hat{y}_i) $$
- 이 손실은 정답 클래스의 확률을 높일수록 감소하며, 잘못된 클래스에 높은 확률을 줄수록 증가한다.
- 전체 데이터셋에 대한 비용 함수는 각 샘플의 크로스 엔트로피를 평균 내어 계산하며, 이를 최소화하도록 경사 하강법을 적용한다.
