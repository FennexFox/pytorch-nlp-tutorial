- 인코딩: 주어진 데이터를 정해진 규칙(code)에 따라 변환하는 것

## 1. 원-핫 인코딩(One-hot encoding)이란?
- 주어진 데이터를 원-핫 벡터로 인코딩하는 방법으로, 아래의 특성을 가짐
    - 선택해야 하는 선택지의 개수만큼의 차원을 가짐
    - 각 선택지의 인덱스에 해당하는 원소에는 1, 나머지 원소는 0을 부여
- `강아지`, `고양이`, `냉장고`를 원-핫 벡터로 표현하면 아래와 같음

|대상|1|2|3|
|--|--|--|--|
|강아지|1|0|0|
|고양이|0|1|0|
|냉장고|0|0|1|

## 2. 원-핫 벡터의 무작위성
- 주어진 데이터를 인코딩하여 벡터로 나타낸 경우, 이 벡터들은 벡터 공간에 표현될 수 있음
- 두 벡터가 서로 유사할 경우 오차 함수의 값이 작아짐: 두 데이터는 서로 유사함
- 독립 관계의 대상들을 인코딩할 경우, 대상들의 거리가 균등해야 함
- 원-핫 벡터는 이러한 성질을 만족함
