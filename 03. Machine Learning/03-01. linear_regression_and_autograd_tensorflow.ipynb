## 5. 텐서플로로 선형 회귀 구현하기

```python
import tensorflow as tf
print(f'TensorFlow version: {tf.__version__}')
tf.random.set_seed(1)
```

```python
x_train = tf.constant([[1.], [2.], [3.]], dtype=tf.float32)
y_train = tf.constant([[2.], [4.], [6.]], dtype=tf.float32)
print(x_train)
print(x_train.shape)
print(y_train)
print(y_train.shape)
```

```python
# 가중치 W와 편향 b를 0으로 초기화
W = tf.Variable(tf.zeros([1, 1]))
b = tf.Variable(tf.zeros([1]))
hypothesis = x_train * W + b
print(hypothesis)
cost = tf.reduce_mean(tf.square(hypothesis - y_train))
print(cost.numpy())
```

```python
optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)
with tf.GradientTape() as tape:
    hypothesis = x_train * W + b
    cost = tf.reduce_mean(tf.square(hypothesis - y_train))
grads = tape.gradient(cost, [W, b])
optimizer.apply_gradients(zip(grads, [W, b]))
print(f'After one step -> W: {W.numpy()}, b: {b.numpy()}, cost: {cost.numpy()}')
```

```python
W = tf.Variable(tf.zeros([1, 1]))
b = tf.Variable(tf.zeros([1]))
optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)
nb_epochs = 5000
for epoch in range(nb_epochs + 1):
    with tf.GradientTape() as tape:
        hypothesis = x_train * W + b
        cost = tf.reduce_mean(tf.square(hypothesis - y_train))
    grads = tape.gradient(cost, [W, b])
    optimizer.apply_gradients(zip(grads, [W, b]))
    if epoch % 100 == 0:
        print(f'Epoch {epoch:4d}/{nb_epochs} W: {W.numpy().item():.3f}, b: {b.numpy().item():.3f} Cost: {cost.numpy():.6f}')
```

## 7. 자동 미분(Autograd) 실습하기

```python
w = tf.Variable(2.0)
with tf.GradientTape() as tape:
    z = 2 * w ** 2 + 5
grad = tape.gradient(z, w)
print('수식을 w로 미분한 값 : {}'.format(grad.numpy()))
```
