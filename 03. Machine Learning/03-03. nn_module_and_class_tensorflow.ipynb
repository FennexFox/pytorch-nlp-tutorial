## 1. 텐서플로로 단순 선형 회귀 구현하기

```python
import tensorflow as tf
print(f'TensorFlow version: {tf.__version__}')
tf.random.set_seed(1)
```

```python
x_train = tf.constant([[1.], [2.], [3.]], dtype=tf.float32)
y_train = tf.constant([[2.], [4.], [6.]], dtype=tf.float32)

model = tf.keras.Sequential([tf.keras.layers.Dense(1, input_shape=(1,))])

optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)
nb_epochs = 2000

for epoch in range(nb_epochs + 1):
    with tf.GradientTape() as tape:
        prediction = model(x_train, training=True)
        cost = tf.reduce_mean(tf.square(prediction - y_train))

    grads = tape.gradient(cost, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))

    if epoch % 100 == 0:
        print(f'Epoch {epoch:4d}/{nb_epochs} Cost: {cost.numpy():.6f}')
```

```python
new_var = tf.constant([[4.]], dtype=tf.float32)
pred_y = model(new_var)

print('훈련 후 입력이 4일 때의 예측값 :', pred_y.numpy())
print('Parameters:', [v.numpy() for v in model.trainable_variables])
```

## 2. 텐서플로로 다중 선형 회귀 구현하기

```python
x_train = tf.constant([[73., 80., 75.],
                       [93., 88., 93.],
                       [89., 91., 90.],
                       [96., 98., 100.],
                       [73., 66., 70.]], dtype=tf.float32)
y_train = tf.constant([[152.],
                       [185.],
                       [180.],
                       [196.],
                       [142.]], dtype=tf.float32)

model = tf.keras.Sequential([tf.keras.layers.Dense(1, input_shape=(3,))])
optimizer = tf.keras.optimizers.SGD(learning_rate=1e-5)
nb_epochs = 2000

for epoch in range(nb_epochs + 1):
    with tf.GradientTape() as tape:
        prediction = model(x_train, training=True)
        cost = tf.reduce_mean(tf.square(prediction - y_train))

    grads = tape.gradient(cost, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))

    if epoch % 100 == 0:
        print(f'Epoch {epoch:4d}/{nb_epochs} Cost: {cost.numpy():.6f}')
```

```python
new_var = tf.constant([[73., 80., 75.]], dtype=tf.float32)
pred_y = model(new_var)
print('훈련 후 입력이 73, 80, 75일 때의 예측값 :', pred_y.numpy())
print('Parameters:', [v.numpy() for v in model.trainable_variables])
```

## 3. tf.keras.Model 클래스로 구현하기 (단순/다중)

```python
class SimpleLinearModel(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.linear = tf.keras.layers.Dense(1)
    def call(self, inputs):
        return self.linear(inputs)

class MultiLinearModel(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.linear = tf.keras.layers.Dense(1)
    def call(self, inputs):
        return self.linear(inputs)

model = MultiLinearModel()
optimizer = tf.keras.optimizers.SGD(learning_rate=1e-5)
nb_epochs = 2000

x_train = tf.constant([[73., 80., 75.],
                       [93., 88., 93.],
                       [89., 91., 90.],
                       [96., 98., 100.],
                       [73., 66., 70.]], dtype=tf.float32)
y_train = tf.constant([[152.],
                       [185.],
                       [180.],
                       [196.],
                       [142.]], dtype=tf.float32)

for epoch in range(nb_epochs + 1):
    with tf.GradientTape() as tape:
        prediction = model(x_train, training=True)
        cost = tf.reduce_mean(tf.square(prediction - y_train))

    grads = tape.gradient(cost, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))

    if epoch % 100 == 0:
        print(f'[Class Model] Epoch {epoch:4d}/{nb_epochs} Cost: {cost.numpy():.6f}')
```
