- 정제(cleaning) : 갖고 있는 코퍼스로부터 노이즈 데이터를 제거
- 정규화(normalization) : 표현 방법이 다른 단어들을 통합

## 1. 표기가 다른 단어들의 통합
- USA와 US는 통합하여 같은 토큰을 부여할 수 있음
- 그러나 US와 us는 품사가 다르며, 서로 다른 토큰이 부여되어야 함

## 2. 대소문자 통합
- 동일한 단어가 문장 내의 위치에 따라 다른 대소문자로 표기되기도 함
- 모든 표현을 소문자로 변환하는 식으로 통합할 수도 있으나, 이는 고유명사 등의 문제를 낳을 수 있음
- 대소문자 통합이 학습된 머신 러닝 시퀀스 모델을 적용할 수 있음

## 3. 불필요한 단어의 제거
- 자연어가 아닌 특수문자 등 뿐 아니라, 분석 목적에 적합하지 않은 자연어 역시 포함됨
- 이는 크게 불용어, 저빈도 단어, 문법 요소 등이 있음

### 3.1 저빈도 단어
- 코퍼스 내에서 너무 적게 등장하여 의미를 부여하기 힘든 경우가 있음

### 3.2 문법 요소
- 영어권 언어의 경우, 2문자 이하의 짧은 단어 중 불용어가 많은 것으로 알려져 있음
- 한국어의 경우, 모아쓰기와 한자 등의 요소로 인해 짧으 단어가 많아 이를 적용하기 어려움

```python
import re
text = "I was wondering if anyone out there could enlighten me on this car."

# 길이가 1~2인 단어들을 정규 표현식을 이용하여 삭제
shortword = re.compile(r'\W*\b\w{1,2}\b')
print(shortword.sub('', text))
```

## 4. 정규 표현식(Regular Expression)
- 노이즈의 특성을 일반화할 수 있다면, 이를 정규 표현식으로 포착하여 제거 가능
