## 1. 토큰화(Tokenization)

```python
en_text = "A Dog Run back corner near spare bedrooms"
```

### 1) spaCy 사용하기

```python
import spacy
spacy_en = spacy.load('en_core_web_sm')
```

```python
def tokenize(en_text):
    return [tok.text for tok in spacy_en.tokenizer(en_text)]
```

```python
print(tokenize(en_text))
```

```python
import nltk
from nltk.tokenize import word_tokenize
print(word_tokenize(en_text))
```

## 2. 단어 집합(Vocabulary) 생성

```python
import urllib.request
import pandas as pd
from konlpy.tag import Mecab
from nltk import FreqDist
import numpy as np
import matplotlib.pyplot as plt
```

```python
urllib.request.urlretrieve("https://raw.githubusercontent.com/e9t/nsmc/master/ratings.txt", filename="ratings.txt")
data = pd.read_table('ratings.txt') # 데이터프레임에 저장
data[:10]
```

```python
print(f'데이터프레임의 크기: {data.shape}')
```

```python
sample_data = data[:100] # 임의로 100개만 저장
```

```python
sample_data['document'] = sample_data['document'].str.replace("[^ㄱ-ㅎㅏ-ㅣ가-힣 ]","", regex=True)
# 한글과 공백을 제외하고 모두 제거
sample_data[:10]
```

```python
# 불용어 정의
stopwords=['의','가','이','은','들','는','좀','잘','걍','과','도','를','으로','자','에','와','한','하다']
```

```python
tokenizer = Mecab()
tokenized=[]
for sentence in sample_data['document']:
    temp = tokenizer.morphs(sentence) # 토큰화
    temp = [word for word in temp if not word in stopwords] # 불용어 제거
    tokenized.append(temp)
```

```python
print("\n".join([" ".join(word) for word in tokenized[:10]]))
```

```python
vocab = FreqDist(np.hstack(tokenized))
print('단어 집합의 크기 : {}'.format(len(vocab)))
```

```python
vocab['재밌']
```

```python
vocab_size = 500
# 상위 vocab_size개의 단어만 보존
vocab = vocab.most_common(vocab_size)
print('단어 집합의 크기 : {}'.format(len(vocab)))
print(vocab[:10]) # 상위 10개 단어 출력: 단어와 등장횟수
```

## 3. 각 단어에 고유한 정수 부여

```python
word_to_index = {word[0] : index + 2 for index, word in enumerate(vocab)}
word_to_index['pad'] = 1
word_to_index['unk'] = 0
```

```python
encoded = []
for line in tokenized: #입력 데이터에서 1줄씩 문장을 읽음
    temp = []
    for w in line: #각 줄에서 1개씩 글자를 읽음
      try:
        temp.append(word_to_index[w]) # 글자를 해당되는 정수로 변환
      except KeyError: # 단어 집합에 없는 단어일 경우 unk로 대체된다.
        temp.append(word_to_index['unk']) # unk의 인덱스로 변환

    encoded.append(temp)
```

```python
print("\n".join([" ".join(map(str, token)) for token in encoded[:10]]))
```

## 4. 길이가 다른 문장들을 모두 동일한 길이로 바꿔주는 패딩(padding)

```python
max_len = max(map(len, encoded))
print('리뷰의 최대 길이 : %d' % max_len)
print('리뷰의 최소 길이 : %d' % min(len(l) for l in encoded))
print('리뷰의 평균 길이 : %f' % (sum(map(len, encoded))/len(encoded)))
plt.hist([len(s) for s in encoded], bins=50)
plt.xlabel('length of sample')
plt.ylabel('number of sample')
plt.show()
```

```python
for line in encoded:
    if len(line) < max_len: # 현재 샘플이 정해준 길이보다 짧으면
        line += [word_to_index['pad']] * (max_len - len(line)) # 나머지는 전부 'pad' 토큰으로 채운다.
```

```python
print('리뷰의 최대 길이 : %d' % max(len(l) for l in encoded))
print('리뷰의 최소 길이 : %d' % min(len(l) for l in encoded))
print('리뷰의 평균 길이 : %f' % (sum(map(len, encoded))/len(encoded)))
```
