## 1. 단어 토큰화(Word Tokenization)
- 언어에는 여러 예외가 있으므로, 엄격히 단어 기준으로 토큰화되지는 않음
- 일반적으로 구두점과 불용어를 제외하고 의미 있는 단위로 분리하여 토큰화함

## 2. 토큰화 중 생기는 선택의 순간
- 구두점, 표현방식 등을 어떻게 처리하는지 여러 방식이 존재
- 해당 데이터를 사용할 용도에 지장이 가지 않는 방식을 선택

```python
from nltk.tokenize import word_tokenize
from nltk.tokenize import WordPunctTokenizer
from tensorflow.keras.preprocessing.text import text_to_word_sequence
```

```python
# nltk의 Word_tokenize 사용
# Don't를 Do + n't로 분리, Jone's를 Jone + 's로 분리
print(f"단어 토큰화1 : {word_tokenize("Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop.")}")
```

```python
# nltk의 WordPunctTokenizer 사용: 구두점을 별도로 분리
# Don't를 Don + ' + t로 분리, Jone's를 Jone + ' + s로 분리
print('단어 토큰화2 :',WordPunctTokenizer().tokenize("Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."))
```

```python
# 케라스의 text_to_word_sequence 사용
# Don't를 don't로 변환, Jone's를 jone's로 변환
print('단어 토큰화3 :',text_to_word_sequence("Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."))
```

## 3. 토큰화에서 고려해야할 사항
- 단순하게 코퍼스에서 구두점을 제외하고 공백 기준으로 잘라내는 것이 아님
- 보다 섬세한 알고리즘으로 여러 경우를 다루어야 함

### 3.1 구두점이나 특수 문자를 보존해야 하는 경우
- 마침표는 `EOS 토큰`이 될 수 있으며, `Ph.D`와 같이 구두점이 표현의 일부인 경우도 있음
- `New York`과 같이 특수문자 등을 포함하나 정제하지 않고 그 자체로 토큰화해야 하는 경우도 있음
- 통화기호 등과 같이 표현의 속성을 보여주는 메타데이터 역할을 하기도 함

### 3.2 표준 토큰화 예제(Penn Treebank Tokenization)
- 하이푼으로 구성된 단어는 하나로 유지한다.
- doesn't와 같이 아포스트로피로 '접어'가 함께하는 단어는 분리해준다.

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()

text = "Starting a home-based restaurant may be an ideal. it doesn't have a food chain or restaurant of their own."
print('트리뱅크 워드토크나이저 :',tokenizer.tokenize(text))
```

## 4. 문장 토큰화(Sentence Tokenization)
- 코퍼스에 포함된 문자열을 문장 단위로 분리하는 작업
- 언어의 국적, 해당 코퍼스에서 특수문자가 사용되는 방식 등에 따라서 여러 규칙들이 가능

```python
from nltk.tokenize import sent_tokenize

text = "His barber kept his word. But keeping such a huge secret to himself was driving him crazy. Finally, the barber went up a mountain and almost to the edge of a cliff. He dug a hole in the midst of some reeds. He looked about, to make sure no one was near."
print(f'문장 토큰화1 : {"\n".join(sent_tokenize(text))}')
```

```python
text = "I am actively looking for Ph.D. students. and you are a Ph.D student."
print(f'문장 토큰화2 : {"\n".join(sent_tokenize(text))}')
```

```python
text = '딥러닝 자연어 처리가 재미있기는 합니다. 그런데 문제라면 영어보다 한국어로 할 때 너무 어렵습니다. 이제 해 보면 알 걸요?'
print(f'한국어 문장 토큰화 : {"\n".join(kss.split_sentences(text))}')
```

## 5. 한국어 토큰화의 어려움
- 한국어는 교착어로서, 실질적 및 형식적 의미를 갖는 여러 형태소가 교착하여 단어를 구성함
- 또한, 한국어는 모아쓰기라는 특징으로 인해 띄어쓰기가 필수적이지 않음
- 따라서 영어 등의 언어처럼 띄어쓰기를 기준으로 토큰화하는 것이 유효하지 않음

## 6. 품사 태깅(Part-of-speech Tagging)
- 외견상 표기는 같으나 품사에 따라서 단어의 의미와 역할이 달라지는 경우가 많음
- 이를 단어 토큰화 과정에서 반영하여 각 단어의 품사를 태깅하는 경우도 있음

## 7. NLTK와 KoNLPy를 이용한 영어, 한국어 토큰화 실습

```python
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag

text = "I am actively looking for Ph.D. students. and you are a Ph.D. student."
tokenized_sentence = word_tokenize(text)

print(f'단어 토큰화 : {tokenized_sentence}')
print(f'품사 태깅 : {pos_tag(tokenized_sentence)}')
```

```python
from konlpy.tag import Okt
from konlpy.tag import Kkma

okt = Okt()
kkma = Kkma()

print('OKT 형태소 분석 :',okt.morphs("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
print('OKT 품사 태깅 :',okt.pos("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
print('OKT 명사 추출 :',okt.nouns("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
```

```python
print('꼬꼬마 형태소 분석 :',kkma.morphs("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
print('꼬꼬마 품사 태깅 :',kkma.pos("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
print('꼬꼬마 명사 추출 :',kkma.nouns("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
```
