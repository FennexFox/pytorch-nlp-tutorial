## 1. 정규 표현식 문법과 모듈 함수
### 1.1 정규 표현식 문법
|특수 문자|설명|
|--|--|
|`.`|한 개의 임의의 문자 (줄바꿈 문자인 `\n`는 제외)|
|`?`|앞의 문자가 1개 이하 존재함|
|`*`|앞의 문자가 0개 이상 존재함|
|`+`|앞의 문자가 1개 이상 존재함|
|`^`|뒤의 문자열로 문자열이 시작됨|
|`$`|앞의 문자열로 문자열이 끝남|
|`{숫자}`|괄호 앞의 문자를 숫자만큼 반복함|
|`{숫자1, 숫자2}`|괄호 앞의 문자를 숫자1 이상 숫자2 이하만큼 반복함|
|`{숫자,}`|괄호 앞의 문자를 숫자 이상만큼 반복함|
|`[ ]`|대괄호 안의 문자들 중 한 개의 문자와 매치함<br>`[amk]`라고 한다면 `a` 또는 `m` 또는 `k` 중 하나라도 존재하면 매치함<br>`[a-z]`와 같이 범위를 지정할 수도 있음<br>`[a-zA-Z]`는 알파벳 전체를 의미하며, 알파벳이 존재하면 매치함|
|`[^문자]`|해당 문자를 제외한 문자를 매치함|
|`\|`|`A\|B`와 같이 쓰이며 `A` 또는 `B`의 의미를 가짐|
### 1.2 이스케이프 문자(Escape Character) 활용
|문자 규칙|설명|
|--|--|
|`\\\`|역 슬래쉬 문자 자체를 의미함|
|`\\d`|모든 숫자를 의미하며, [0-9]와 의미가 동일|
|`\\D`|숫자를 제외한 모든 문자를 의미하며, [^0-9]와 의미가 동일|
|`\\s`|공백을 의미하며, [ \t\n\r\f\v]와 의미가 동일|
|`\\S`|공백을 제외한 문자를 의미하며, [^ \t\n\r\f\v]와 의미가 동일|
|`\\w`|문자 또는 숫자를 의미하며, [a-zA-Z0-9]와 의미가 동일|
|`\\W`|문자 또는 숫자가 아닌 문자를 의미하며, [^a-zA-Z0-9]와 의미가 동일|
### 1.3 정규 표현식 모듈 함수
|모듈 함수|설명|
|--|--|
|`re.compile()`|정규표현식을 컴파일하여 파이썬 객체로 만듦|
|`re.search()`|문자열 전체에 대해서 정규표현식과 매치되는지를 검색|
|`re.match()`|문자열의 처음이 정규표현식과 매치되는지를 검색|
|`re.split()`|정규 표현식을 기준으로 문자열을 분리하여 리스트로 리턴|
|`re.findall()`|문자열에서 정규 표현식과 매치되는 모든 경우의 문자열을 찾아서 리스트로 리턴<br> 매치되는 문자열이 없다면 빈 리스트를 리턴|
|`re.finditer()`|문자열에서 정규 표현식과 매치되는 모든 경우의 문자열에 대한 이터레이터 객체를 리턴|
|`re.sub()`|문자열에서 정규 표현식과 일치하는 부분에 대해서 다른 문자열로 대체|

## 2. 정규 표현식 실습

```python
import re
```

### 2.1 `.`기호

```python
r = re.compile("a.c")
r.search("kkk") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abc")
```

### 2.2 `?`기호

```python
r = re.compile("ab?c")
r.search("abbc") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abc")
```

```python
r.search("ac")
```

### 2.3 `*`기호

```python
r = re.compile("ab*c")
r.search("a") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("ac")
```

```python
r.search("abc") 
```

```python
r.search("abbbbc")
```

### 2.4 `+`기호

```python
r = re.compile("ab+c")
r.search("ac") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abc")
```

```python
r.search("abbbbc")
```

### 2.5 `^`기호

```python
r = re.compile("^ab")
r.search("bbc") # 아무런 결과도 출력되지 않는다.
r.search("zab") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abz")
```

### 2.6 `{숫자}` 기호

```python
r = re.compile("ab{2}c")
r.search("ac") # 아무런 결과도 출력되지 않는다.
r.search("abc") # 아무런 결과도 출력되지 않는다.
r.search("abbbbbc") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abbc")
```

### 2.7 `{숫자1, 숫자2}` 기호

```python
r = re.compile("ab{2,8}c")
r.search("ac") # 아무런 결과도 출력되지 않는다.
r.search("abc") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abbc")
```

```python
r.search("abbbbbbbbc")
```

```python
r.search("abbbbbbbbbc") # 아무런 결과도 출력되지 않는다.
```

### 2.8 `{숫자,}` 기호

```python
r = re.compile("a{2,}bc")
r.search("bc") # 아무런 결과도 출력되지 않는다.
r.search("aa") # 아무런 결과도 출력되지 않는다.
r.search("aaaaaaaabc") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("aabc")
```

### 2.9 `[]` 기호

```python
r = re.compile("[abc]") # [abc]는 [a-c]와 같다.
r.search("zzz") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("a")
```

```python
r.search("aaaaaaa")      
```

```python
r.search("baac") 
```

```python
r = re.compile("[a-z]")
r.search("AAA") # 아무런 결과도 출력되지 않는다.
r.search("111") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("aBC")
```

### 2.10 `[^문자]` 기호

```python
r = re.compile("[^abc]")

# 아래의 세 코드는 아무런 결과도 출력되지 않는다.
r.search("a")
r.search("ab") 
r.search("b")
```

```python
r.search("d")
```

```python
r.search("1")
```

## 3. 정규 표현식 모듈 함수 예제


### 3.1 `re.match()` 와 `re.search()`의 차이

```python
r = re.compile("ab.")
r.match("kkkabc")  #아무런 결과도 출력되지 않는다.
```

```python
r.search("kkkabc")  
```

```python
r.match("abckkk")  
```

### 3.2 `re.split()`

```python
text = "사과 딸기 수박 메론 바나나"
re.split(" ",text)
```

```python
text="""사과
딸기
수박
메론
바나나"""

re.split("\n",text)
```

```python
text = "사과+딸기+수박+메론+바나나"

re.split("\+",text)
```

### 2.3 `re.findall()`

```python
text = """이름 : 김철수
전화번호 : 010 - 1234 - 1234
나이 : 30
성별 : 남"""

re.findall("\\d+",text)
```

```python
re.findall("\\d+", "문자열입니다.")
```

### 2.4 `re.sub()`

```python
text = "Regular expression : A regular expression, regex or regexp[1] (sometimes called a rational expression)[2][3] is, in theoretical computer science and formal language theory, a sequence of characters that define a search pattern."

preprocessed_text = re.sub('[^a-zA-Z]', ' ', text)
print(preprocessed_text)
```

## 4. 정규 표현식 텍스트 전처리 예제

```python
text = """100 John    PROF
101 James   STUD
102 Mac   STUD"""  

re.split('\\s+', text)
```

```python
re.findall('\\d+',text)
```

```python
re.findall('[A-Z]',text)
```

```python
re.findall('[A-Z]{4}',text)
```

```python
re.findall('[A-Z][a-z]+',text)
```

## 5. 정규 표현식을 이용한 토큰화

```python
from nltk.tokenize import RegexpTokenizer

text = "Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop"

tokenizer1 = RegexpTokenizer("[\\w]+")
tokenizer2 = RegexpTokenizer("[\\s]+", gaps=True)

print(tokenizer1.tokenize(text))
print(tokenizer2.tokenize(text))
```
