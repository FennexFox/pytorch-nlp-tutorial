## 1. 단어의 표현 방법
### 1.1 이산 표현(Discrete Representation)
- 문장 내에서 다른 단어와의 관계를 표현하지 않고, 해당 단어 자체의 특성만 표현
- 각각의 단어는 서로 다른 차원에 존재하며 서로간에 연속되는 특성이 없음
- 구조가 단순하다는 장점이 있으나, 단어의 의미나 유사성을 표현할 수 없다
- 텐서의 구조가 희소(Sparse)함
### 1.2 연속 표현(Continous Representation)
- 특정 단어를 동일한 잠재 공간에 벡터로 표현하며, 벡터의 여러 차원이 함께 단어의 의미를 표현
- 비슷한 단어의 벡터는 잠재 공간에서 가까이 존재하며, 벡터 연산도 가능함
- 벡터의 차원 각각에 대응되는 의미는 없음
- 텐서의 구조가 밀집(Dense)됨
