## 2. 시그모이드 함수(Sigmoid function)


```python
import numpy as np  # 넘파이 사용
import matplotlib.pyplot as plt  # 맷플롯립 사용
```

```python
def sigmoid(x):  # 시그모이드 함수 정의
    return 1 / (1 + np.exp(-x))
```

```python
fig, ax = plt.subplots(figsize=(8, 6))  # 그래프 크기 설정
x = np.linspace(-10, 10, 100)  # -10에서 10까지 100개의 점 생성
y = sigmoid(x)  # 시그모이드 함수 적용
ax.plot(x, y, label='Sigmoid Function', color='blue')  # 그래프 그리기
ax.axhline(0, color='black', linewidth=0.5, ls='-')  # x축 추가
ax.axvline(0, color='black', linewidth=0.5, ls='-')  # y축 추가
ax.grid(True, linestyle='--', linewidth=0.5)  # 격자 추가
ax.set_title('Sigmoid Activation Function')  # 그래프 제목 설정
ax.set_xlabel('Input')  # x축 라벨 설정
ax.set_ylabel('Output')  # y축 라벨 설정
ax.legend()  # 범례 추가
plt.show()
```

```python
x = np.linspace(-10, 10, 100)  # -10에서 10까지 100개의 점 생성
y1 = sigmoid(0.5 * x)
y2 = sigmoid(x)
y3 = sigmoid(2 * x)

fig, ax = plt.subplots(figsize=(8, 6))  # 그래프 크기 설정
ax.plot(x, y1, label='W=0.5', color='r', linestyle='--')  # W의 값이 0.5일 때
ax.plot(x, y2, label='W=1', color='g')  # W의 값이 1일 때
ax.plot(x, y3, label='W=2', color='b', linestyle='--')  # W의 값이 2일 때
ax.axhline(0, color='black', linewidth=0.5, ls='-')  # x축 추가
ax.axvline(0, color='black', linewidth=0.5, ls='-')  # y축 추가
ax.grid(True, linestyle='--', linewidth=0.5)  # 격자 추가
ax.set_title('Sigmoid Function')
ax.set_xlabel('Input')  # x축 라벨 설정
ax.set_ylabel('Output')  # y축 라벨 설정
ax.legend()  # 범례 추가
plt.show()
```

## 4. 텐서플로우로 로지스틱 회귀 구현하기


```python
import tensorflow as tf
tf.random.set_seed(1)
```

```python
x_data = [[1, 2], [2, 3], [3, 1], [4, 3], [5, 3], [6, 2]]
y_data = [[0], [0], [0], [1], [1], [1]]
x_train = tf.constant(x_data, dtype=tf.float32)
y_train = tf.constant(y_data, dtype=tf.float32)

print(x_train.shape)
print(y_train.shape)
```

```python
W = tf.Variable(tf.zeros((2, 1)))  # 크기는 2 x 1
b = tf.Variable(tf.zeros((1,)))

logits = tf.matmul(x_train, W) + b
hypothesis = tf.sigmoid(logits)  # 시그모이드 함수 적용

print(hypothesis)  # 예측값인 H(x) 출력
```

```python
losses = -(y_train * tf.math.log(hypothesis) + (1 - y_train) * tf.math.log(1 - hypothesis))
cost = tf.reduce_mean(losses)

print(losses)
print(cost)
```

```python
bce = tf.keras.losses.BinaryCrossentropy(from_logits=False)
bce_cost = bce(y_train, hypothesis)
bce_cost
```

```python
x_data = [[1, 2], [2, 3], [3, 1], [4, 3], [5, 3], [6, 2]]
y_data = [[0], [0], [0], [1], [1], [1]]
x_train = tf.constant(x_data, dtype=tf.float32)
y_train = tf.constant(y_data, dtype=tf.float32)

W = tf.Variable(tf.zeros((2, 1)))
b = tf.Variable(tf.zeros((1,)))
optimizer = tf.keras.optimizers.SGD(learning_rate=1.0)

nb_epochs = 1000
for epoch in range(nb_epochs + 1):
    with tf.GradientTape() as tape:
        logits = tf.matmul(x_train, W) + b
        hypothesis = tf.sigmoid(logits)
        cost = tf.reduce_mean(-(y_train * tf.math.log(hypothesis) + (1 - y_train) * tf.math.log(1 - hypothesis)))

    grads = tape.gradient(cost, [W, b])
    optimizer.apply_gradients(zip(grads, [W, b]))

    if epoch % 100 == 0:
        print(f'Epoch {epoch:4d}/{nb_epochs} Cost: {cost.numpy():.6f}')
```

```python
logits = tf.matmul(x_train, W) + b
hypothesis = tf.sigmoid(logits)
print(hypothesis)

prediction = hypothesis >= 0.5
print(prediction)

print(W)
print(b)
```
